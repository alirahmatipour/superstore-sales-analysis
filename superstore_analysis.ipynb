## Data Loading

```python
# Load the dataset and preview the first rows
import pandas as pd
df = pd.read_csv('Sample-Superstore.csv', encoding='latin1')
df.head()
```

```python
# Check the number of rows and columns
df.shape
```

## Data Cleaning

```python
# Inspect column types and check for missing values
df.info()
```

```python
# Check for missing values in each column
df.isnull().sum()
```

```python
# Check for duplicate rows
df.duplicated().sum()
```

```python
# Convert date columns to proper datetime format for time-based analysis
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df.dtypes
```

## 1. Category & Sub-Category Profitability Analysis

```python
# Calculate total sales and profit per category, then add a profit margin column
category_summary = df.groupby('Category')[['Sales', 'Profit']].sum()
category_summary['Profit Margin'] = category_summary['Profit'] / category_summary['Sales']
category_summary
```

```python
# Calculate total sales and profit per sub-category, add profit margin, sort by margin
subcategory_summary = df.groupby('Sub-Category')[['Sales', 'Profit']].sum()
subcategory_summary['Profit Margin'] = subcategory_summary['Profit'] / subcategory_summary['Sales']
subcategory_summary.sort_values('Profit Margin')
```

**Finding:** Furniture has a much lower profit margin (2.5%) compared to 
Office Supplies and Technology (both ~17%), despite similar sales volume. 
Digging into sub-categories reveals the cause: Tables (-8.6% margin) and 
Bookcases (-3.0% margin) are actually losing money, dragging down the 
entire Furniture category.


## 2. Regional & State Profit Analysis

```python
# Calculate total sales and profit per region, add profit margin, sort by profit
region_summary = df.groupby('Region')[['Sales', 'Profit']].sum()
region_summary['Profit Margin'] = region_summary['Profit'] / region_summary['Sales']
region_summary.sort_values('Profit')
```

```python
# Calculate total sales and profit per state, add profit margin, sort by profit
state_summary = df.groupby('State')[['Sales', 'Profit']].sum()
state_summary['Profit Margin'] = state_summary['Profit'] / state_summary['Sales']
state_summary.sort_values('Profit').head(10)
```

```python
# Calculate total sales and profit per city (grouped with state to avoid mixing same-name cities), showing the 10 worst-performing cities
city_summary = df.groupby(['State', 'City'])[['Sales', 'Profit']].sum()
city_summary['Profit Margin'] = city_summary['Profit'] / city_summary['Sales']
city_summary.sort_values('Profit').head(10)
```

**Finding:** Texas has the largest absolute loss (-$25,729), while Ohio has 
the worst relative performance (-21.7% profit margin). Drilling down to the 
city level reveals that Ohio's poor performance is driven almost entirely by 
one city, Lancaster, which has an extreme -87.2% profit margin — showing 
that state-level losses can be concentrated in just a few locations rather 
than spread evenly.


## 3. Discount vs Profitability Analysis

```python
# Calculate average discount and profit for each sub-category, sorted by discount (highest first)
discount_analysis = df.groupby('Sub-Category')[['Discount', 'Profit']].mean()
discount_analysis.sort_values('Discount', ascending=False)
```


**Finding:** Tables and Bookcases receive high average discounts (26% and 
21%) and show negative average profit, confirming that heavy discounting 
hurts profitability for these sub-categories. However, this relationship 
isn't strictly linear — Binders receives the highest discount (37%) but 
remains profitable, suggesting the impact of discounting depends on each 
product's baseline profit margin.
